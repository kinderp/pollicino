Dal logit al primo modello neurale byte-level, senza framework ML.

```python
import math
def softmax(z):
    m=max(z); e=[math.exp(x-m) for x in z]; s=sum(e); return [x/s for x in e]
softmax([1,2,3])
```

```python
p=softmax([0,3,0])
loss=-math.log2(p[1])
p, loss
```
