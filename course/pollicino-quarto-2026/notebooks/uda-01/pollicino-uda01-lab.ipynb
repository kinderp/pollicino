## 2. Hash e collisioni

SHA-256 produce 256 bit. Per rendere visibili le collisioni in classe, tronchiamo artificialmente il digest a 8 bit.


```python
import hashlib

def short_hash(data: bytes, bits: int = 8) -> int:
    digest = int.from_bytes(hashlib.sha256(data).digest(), "big")
    return digest >> (256 - bits)

seen = {}
for i in range(257):
    candidate = f"candidate-{i}".encode()
    h = short_hash(candidate, 8)
    if h in seen:
        print("collisione:", seen[h], candidate, "->", h)
        break
    seen[h] = candidate
```

**Perché 257 tentativi bastano a garantire almeno una collisione con 8 bit?** Collega la risposta al principio dei cassetti.


## 3. Quanto è sorprendente un evento?

Usiamo `I(x) = -log2(p)`. Se la probabilità si dimezza, l'informazione aumenta di 1 bit.


```python
import math

for p in [1, 1/2, 1/4, 1/16, 1/256]:
    print(f"p={p:10.6f}  I={-math.log2(p):5.2f} bit")
```

## 4. Entropia empirica di una sequenza

L'entropia zero-order usa solo le frequenze dei byte, senza considerare il contesto.


```python
from collections import Counter

def entropy(data: bytes) -> float:
    if not data:
        return 0.0
    n = len(data)
    return -sum((c/n) * math.log2(c/n) for c in Counter(data).values())

samples = {
    "ripetitivo": b"A" * 100,
    "due simboli": b"AB" * 50,
    "quattro simboli": b"ABCD" * 25,
    "tutti i byte": bytes(range(256)),
}
for name, sample in samples.items():
    print(f"{name:14s}: {entropy(sample):.4f} bit/byte")
```
