# POLLICINO — UDA 4: Transformer da zero
Dal contesto causale al Tiny Byte Transformer.

```python
def causal_mask(n):
    return [[j <= i for j in range(n)] for i in range(n)]
for row in causal_mask(6):
    print(' '.join('1' if x else '0' for x in row))
```

## Scaled dot-product attention
Il fattore $1/\sqrt{d_k}$ evita score troppo grandi quando cresce la dimensione.

```python
import math
def softmax(xs):
    m=max(xs); e=[math.exp(x-m) for x in xs]; s=sum(e); return [x/s for x in e]
q=[1.0,0.0]; keys=[[1.0,0.0],[0.0,1.0]]
scores=[sum(a*b for a,b in zip(q,k))/math.sqrt(2) for k in keys]
print(scores, softmax(scores))
```

## Invariante fondamentale
Cambiare il futuro non deve cambiare l'output del prefisso. Questo test contro il future leakage resterà valido anche nelle implementazioni PyTorch e MLX.
